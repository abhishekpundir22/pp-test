

If we list all the natural numbers below 10 that are multiples of 3 or 5, we get 3, 5, 6 and 9. The sum of these multiples is 23.

Find the sum of all the multiples of 3 or 5 below 1000.


```python
def sum_multiples(limit: int, divisors=(3, 5)) -> int:
    """Sum of all natural numbers below `limit` divisible by any of `divisors`."""
    return sum(i for i in range(limit) if any(i % d == 0 for d in divisors))

assert sum_multiples(10) == 23 # This is a small unit test case to check if code works correctly on test case
print(sum_multiples(1000))  # 233168
```
